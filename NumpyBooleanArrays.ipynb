##**Boolean arrays**
_(Also called masks)_

```python
a = np.arange(4)
```

```python
a
```

```python
a[[0, -1]] #lista de índices
```

```python
a[[True, False, True, False]]
```

```python
a >= 2
```

```python
a[a >=2]
```

```python
a.mean()
```

```python
a[a > a.mean()]
```

```python
a[~(a > a.mean())] #valores de a que não são maiores que a média
```

```python
a[(a==0 | (a==1))]
```

```python
a[(a <= 2) & (a % 2 ==0)]
```
