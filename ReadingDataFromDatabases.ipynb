#**Reading data from relational databases**

```python
import pandas as pd
```

```python
import sqlite3
```

##**Using pandas `read_sql` method*
It will return a DataFrame.

```python
conn = sqlite3.connect("/content/Chinook_Sqlite.sqlite")

df = pd.read_sql("SELECT * FROM Employee LIMIT 5;", conn)
df
```

```python
df = pd.read_sql('SELECT * FROM Employee;', conn,
                 index_col='EmployeeId',
                 parse_dates=['BirthDate', 'HireDate'])
```

```python
df.head()
```

```python
df.info()
```

```python
df['ReportsTo'].isna().sum()
```

```python
df['ReportsTo'].mean()
```

```python
df['ReportsTo'] > 1.75
```

```python
df['City'] = df['City'].astype('category')
```

```python
df.info()
```

##**Using pandas `read_sql_query` method**
It turns out that the read_sql method is just a wrapper around read_sql_query and read_sql_table.
We can get the sam eresult using read_sql_query method:

```python
conn = sqlite3.connect("/content/Chinook_Sqlite.sqlite")
```

```python
df = pd.read_sql_query('SELECT * FROM Employee LIMIT 5;', conn)
```

```python
df.head()
```

```python
df = pd.read_sql_query('SELECT * FROM Employee;', conn,
                 index_col='EmployeeId',
                 parse_dates=['BirthDate', 'HireDate'])
```

```python
df.head()
```
