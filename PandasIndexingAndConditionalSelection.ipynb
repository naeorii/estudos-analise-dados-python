```python
import numpy as np
import pandas as pd
```

##**Indexing**

```python
# In millions
g7_pop = pd.Series([35.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523])
```

```python
g7_pop.name = 'G7 population in millions'
```

```python
g7_pop
```

```python
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States',
]
```

```python
g7_pop['Canada']
```

```python
g7_pop['Japan']
```

Numeric positions can also be used, witch the `iloc` atribute

```python
g7_pop.iloc[0]
```

```python
g7_pop.iloc[-1]
```

Selecting multiple elements at once:

```python
g7_pop[['Italy', 'France']]  # the result is another Series
```

```python
g7_pop.iloc[[0, 1]]
```

In Pandas the upper limit is also included

```python
g7_pop['Canada': 'Italy']
```

---

##**Conditional selection (boolean arrays)**

```python
g7_pop
```

```python
g7_pop > 70  # o numpy verifica cada valor e retorna True ou False
```

```python
g7_pop[g7_pop > 70]  # mostra os valores que atendem a condição
```

```python
g7_pop.mean()
```

```python
g7_pop[g7_pop > g7_pop.mean()]
```

```python
g7_pop.std()
```

```python
g7_pop[(g7_pop > g7_pop.mean() - g7_pop.std() / 2) | (g7_pop > g7_pop.mean() + g7_pop.std() / 2)]
```

##**Boolean arrays**
(Work in the same way as NumPy)

```python
g7_pop
```

```python
g7_pop > 80
```

```python
g7_pop[g7_pop > 80]
```

```python
g7_pop[(g7_pop > 80) | (g7_pop < 40)]
```

```python
g7_pop[(g7_pop > 80) & (g7_pop < 200)]
```
