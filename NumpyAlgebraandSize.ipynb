##**Size of objects in Memory**

###**ints, floats**

```python
# An integer in python is > 24bytes
sys.getsizeof(1)
```

```python
# longs are even larger
sys.getsizeof(10**100)
```

```python
# Numpy size is much smaller
np.dtype(int).itemsize
```

```python
np.dtype(float).itemsize
```

###**Lists are even larger**

```python
# A one-element list
sys.getsizeof([1])
```

```python
# An array of one element in numpy
np.array([1]).nbytes
```

###**And performance is also important**

```python
l = list(range(100))
```

```python
a = np.arange(100)
```

```python
%time np.sum(a ** 2)
```

```python
%time sum([x ** 2 for x in l])
```

##**Useful Numpy functions**

`random`

```python
np.random.random(size=2) # acessa o módulo random do NumPy e usa a função random() para gerar 2 números aleatórios entre 0 e 1
```

```python
np.random.normal(size=2) # gera 2 números aleatórios seguindo uma distribuição normal, com média 0 e desvio padrão 1
```

```python
np.random.rand(2, 4) # gera uma matriz com 2 linhas e 4 colunas, preenchida com números decimais aleatórios entre 0 e 1.
```

`arange`



```python
np.arange(10) # cria um array com os números de 0 a 9
```

```python
np.arange(5, 10)
```

```python
np.arange(0, 1, .1)
```

`reshape`

```python
np.arange(10).reshape(2, 5)
```

```python
np.arange(10).reshape(5, 2)
```
