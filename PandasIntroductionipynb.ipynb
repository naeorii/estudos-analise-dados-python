##**Pandas Series**

We'll start analyzing the "The Group of Seven".
We'll use a panda.Series object



```python
# In millions
g7_pop = pd.Series([35.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523])
```

```python
g7_pop
```

Series can have a `name`, to better document the purpose of the series.

```python
g7_pop.name = 'G7 population in millions'
```

```python
g7_pop
```

Series are similar to numpy arrays.

```python
g7_pop.dtype
```

```python
g7_pop.values
```

A series has a `index`

```python
g7_pop[0]
```

```python
g7_pop[1]
```

```python
g7_pop.index
```

```python
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States',
]
```

```python
g7_pop
```

We can create Series out of dictionaries

```python
pd.Series({
    'Canada': 35.467,
    'France': 63.951,
    'Germany': 80.94,
    'Italy': 60.665,
    'Japan': 127,
    'United Kingdom': 64.511,
    'United States': 318.523
}, name='G7 Population in millions')
```

```python
pd.Series(
    [35.467, 63.951, 80.94, 60.665, 127, 64.511, 318.523],
    index=['Canada', 'France', 'Germany', 'Italy', 'Japan', 'United Kingdom',
       'United States'],
    name='G7 Population in millions'
)
```

Create Series out of other Series

```python
pd.Series(g7_pop, index=['France', 'Germany', 'Italy', 'Spain'])
```
