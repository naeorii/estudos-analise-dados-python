##**Parsing raw HTML strings**
read_html() will read HTML tables from a URL, a file-like object or a raw string containing HTML, and return a list of DataFrame objects.

```python
html_string = """
<table>
        <thead>
            <tr>
                <th>Sequence</th>
                <th>Country</th>
                <th>Population</th>
                <th>Updated</th>
            </tr>
        </thead>
        <tbody>
            <tr>
                <td>1</td>
                <td>China</td>
                <td>1,439,323,776</td>
                <td>1-Dec-2020</td>
            </tr>
            <tr>
                <td>2</td>
                <td>India</td>
                <td>1,380,004,385</td>
                <td>1-Dec-2020</td>
            </tr>
            <tr>
                <td>3</td>
                <td>United States</td>
                <td>331,002,651</td>
                <td>1-Dec-2020</td>
            </tr>
        </tbody>
    </table>
    """
```

```python
from IPython.core.display import display, HTML
display(HTML(html_string))
```

```python
dfs = pd.read_html(html_string)
```

The read_html just returner one `DataFrame` object:

```python
len(dfs)
```

One website can have multiple tables.

```python
df = dfs[0]

df
```

```python
df.shape
```

```python
df.loc[df['Country'] == 'China']
```

##**Defining header**

Pandas will automatically find the header to use thanks to the tag.

But in many cases we'll find wrong or incomplete tables that make the `read_html` method parse the tables in wrong way whithout the proper headers.

To fiz them we can use the `header` parameter.

```python
html_string = """
<table>
        <thead>
            <tr>
                <th>Sequence</th>
                <th>Country</th>
                <th>Population</th>
                <th>Updated</th>
            </tr>
        </thead>
        <tbody>
            <tr>
                <td>1</td>
                <td>China</td>
                <td>1,439,323,776</td>
                <td>1-Dec-2020</td>
            </tr>
            <tr>
                <td>2</td>
                <td>India</td>
                <td>1,380,004,385</td>
                <td>1-Dec-2020</td>
            </tr>
            <tr>
                <td>3</td>
                <td>United States</td>
                <td>331,002,651</td>
                <td>1-Dec-2020</td>
            </tr>
        </tbody>
    </table>
    """
```

```python
pd.read_html(html_string)[0]
```

We'll need to pass the row number to use as header using the `header` parameter.

```python
pd.read_html(html_string, header=0)[0]
```

```python
import pandas as pd
import requests
from io import StringIO

html_url = "https://en.wikipedia.org/wiki/The_Simpsons"

headers = {"User-Agent": "Mozilla/5.0"}

r = requests.get(html_url, headers=headers)
r.raise_for_status()


wiki_tables = pd.read_html(StringIO(r.text), header=0)


simpsons = wiki_tables[2].copy()

simpsons.head()
```

```python
simpsons.drop([0, 1], inplace=True)
```

```python
simpsons.set_index('Season', inplace=True)
```

Wich season  has the lowest number of episodes?

```python
simpsons['No. of episodes'].unique()
```

```python
simpsons = simpsons.loc[simpsons['No. of episodes'] != 'TBA']
```

```python
min_season = simpsons['No. of episodes'].min()

min_season
```

```python
simpsons.loc[simpsons['No. of episodes'] == min_season]
```
