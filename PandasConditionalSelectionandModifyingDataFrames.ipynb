```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    'Population': [35.467, 63.951, 80.94, 60.665, 127.061, 64.511, 318.523],
    'GPD': [
        1785387,
        2833687,
        3874437,
        2167744,
        4602367,
        2950039,
        17348075
    ],
    'Surface Area': [
        9984670,
        640679,
        357114,
        301336,
        377930,
        242495,
        9525067
    ],
    'HDI': [
        0.913,
        0.888,
        0.916,
        0.873,
        0.891,
        0.907,
        0.915
    ],
    'Continent': [
        'America',
        'Europe',
        'Europe',
        'Europe',
        'Asia',
        'Europe',
        'America'
    ]
}, columns=['Population', 'GPD', 'Surface Area', 'HDI', 'Continent'])
```

```python
df.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States',
]
```

```python
df
```

##**Conditional selection (boolean arrays)**

```python
df['Population'] > 70
```

```python
df.loc[df['Population'] > 70]
```

```python
df.loc[df['Population'] > 70, 'Population']
```

```python
df.loc[df['Population'] > 70, ['Population', 'GPD']]
```

##**Modifying DataFrames**

###Adding a new column

```python
langs = pd.Series (
    ['French', 'German', 'Italian'],
    index = ['France', 'Germany', 'Italy'],
    name = 'Language'
)
```

```python
df['Language'] = langs
```

```python
df
```

---

Replacing values per column

```python
df['Language'] = 'English'
```

```python
df
```

Renaming columns

```python
# mudanças não permanentes
```

```python
df.rename(
    columns = {
        'HDI': 'Human Development Index',
        'Anual Popcorn Consumption': 'APC'
    }, index = {
        'United States': 'USA',
        'United Kingdom': 'UK',
        'Argentina': 'AR'
    })
```

```python
df.rename(index = str.upper)
```

```python
df.rename(index = lambda x: x.lower())
```
