```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame ({
        'Sex': ['M', 'F', 'F', 'D', '?'],
        'Age': [29, 30, 24, 290, 25],
})

df
```

How can we clean the not-misisng values but clearly invalid values?

##**Finding Unique Values**

We start by analizing the variety of value spresent.For that, we use the `unique()` method.

```python
df['Sex'].unique()
```

```python
df['Sex'].value_counts()
```

You can use the `replace` function to replace values like `'D'` and `'?'`.

```python
df['Sex'].replace({'D': 'F', 'N': 'M'})
```

If you have many columns to replace, you could apply it at "DataFrame" level

```python
df.replace({
    'Sex': {
        'D' : 'F',
        'N' : 'M'
    },
    'Age': {
        290 : 29
    }
})
```

What if you'd like to remove all the extra 0s from the ages columns? (example, 150 > 15, 490 > 49).

We can combine boolean selection with the operation:

```python
df[df['Age'] > 100]
```

And we can now just divide by 10:

```python
# selecione, na coluna Age, somente as linhas em que Age > 100.

# df['Age'] > 100 → escolhe somente as linhas em que Age é maior que 100
# 'Age' → escolhe somente a coluna Age
df.loc[df['Age'] > 100, 'Age'] = df.loc[df['Age'] > 100, 'Age'] / 10
```
