##**Indexing, Selection and Slicing**
Individual columns can be selected with regular indexing. Each column is represented as a `Series`.

```python
df
```

```python
df['Population'] #coluna individual selecionada
```

Note that the index of the returned Series is the same as the DataFrame one. And its name is the name of the column. If you're working on a notebook and want to see a more DataFrame-like format, you can use the `to_frame()` method.

```python
df['Population'].to_frame()
```

Multiple columns can also be selected.

```python
df[['Population', 'GPD']]  # nesse caso o resultado é outro DataFrame.
```

Slicing works differently, it acts at 'row level' and can be counter intuitive.

```python
df[1:3]  # faz um fatiamento por posição e retorna as linhas que estão nas posições 1 e 2.
```

Row level selection works better with `loc` and `iloc.`
`loc` selects rows matching the given index:

```python
df.loc['Italy'] # seleciona uma linha pelo rótulo do índice
```

```python
df.iloc[-1]
```

```python
df.loc['France':'Italy']
```

As a second "argument" you can pass the column(s) you'd like to select.

```python
df.loc['France':'Italy', 'Population']
```

```python
df.loc['France':'Italy', ['Population', 'GPD']]
```

iloc works with the (numeric) "position" of the index.

```python
df
```

```python
df.iloc[0]
```

```python
df.iloc[-1]
```

```python
df.iloc[[0, 1, -1]]
```

```python
df.iloc[1:3, 3]
```

```python
df.iloc[1:3,[0, 3]]  #index e colunas desejadas
```

```python
df.iloc[1:3, 1:3]
```

Recomenda-se sempre usar `loc` e `iloc`.
