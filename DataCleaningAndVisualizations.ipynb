##**More Visualizations**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

##**Global API**

Matplotlib's default pyplot API has a global, MATLAB-style interface, as we've already seen:

```python
x = np.arange(-10, 11)
```

```python
x
```

```python
plt.figure(figsize=(12, 6))  # cria a área onde o gráfico será desenhado

plt.title('My Nice Plot')    # título

plt.plot(x, x**2)
plt.plot(x, -1 * (x ** 2))

# plt → matplotlib.pyplot
# plot() → desenha um gráfico
# primeiro argumento x → valores do eixo horizontal
# segundo argumento → valores correspondentes do eixo vertical y
```

```python
plt.plot([-10, 0, 10], [-50, -50, -50])
```

```python
# prepara uma figura com espaço para dois gráficos lado a lado
plt.figure(figsize=(12, 6))
plt.suptitle('My Nice Plot')  # título da figura inteira

plt.subplot(1, 2, 1) # linhas, colunas, posição
# tudo abaixo vai para o primeiro painel, até selecionar outro subplot
plt.plot(x, x**2)   #x no eixo horizontal e x^2 no eixo vertical
plt.plot([0, 0, 0], [-10, 0, 100])  #arrays com valores de x e y
plt.legend(['X^2', 'Vertical Line'])
plt.xlabel('X')
plt.ylabel('X Squared')


plt.subplot(1, 2, 2)
plt.plot(x, -1 * (x ** 2))
plt.plot([-10, 0, 10], [-50, -50, -50])  # cria uma linha horizontal em y = -50, ligando pontos com valores de x diferentes
plt.legend(['-X^2', 'Horizontal Line'])
plt.xlabel('X')
plt.ylabel('-X Squared')
```

##**OOP Interface**

```python
fig, axes = plt.subplots(figsize=(12, 6))
```

```python
axes.plot(
  x, x ** 2, color='red', linewidth=3,
  marker='o', markersize=8, label='X^2'
)

axes.plot(x, -1 * (x ** 2), 'b--', label='-X^2')

axes.set_xlabel('X')
axes.set_ylabel('X Squared')

axes.set_title('My Nice Plot')

axes.legend()

fig
```

```python
fig, axes = plt.subplots(figsize= (12,6))

axes.plot(x, x + 0, linestyle="solid")
axes.plot(x, x + 1, linestyle="dashed")
axes.plot(x, x + 2, linestyle="dashdot")
axes.plot(x, x + 3, linestyle="dotted")

axes.set_title("My nice plot")
```

##**Other types of plots**
###**Figures and subfigures

When we call the `subplots() `functions we get a tuple containing a `Figure `and a `axes `element.

```python
plot_objetcs = plt.subplots()

fig, ax = plot_objetcs

ax.plot([1, 2, 3], [1, 2, 3])

plot_objetcs
```

We can also define how many elements we want inside our figure. To do that we can set the `nrows` and `ncols `params.

```python
plot_objetcs = plt.subplots(nrows=2, ncols=2, figsize=(14, 6))

fig, ((ax1, ax2), (ax3, ax4)) = plot_objetcs

plot_objetcs
```

```python
ax4.plot(np.random.randn(50), c='yellow')
ax1.plot(np.random.randn(50), c='red', linestyle='--')
ax2.plot(np.random.randn(50), c='blue', linestyle=":")
ax3.plot(np.random.randn(50), c='green', marker="o", linewidth=3.0)
```

###The` subplot2grid` command

There is another way to make subplots usind grid-like format:

```python
plt.figure(figsize=(14, 6))

#plt.subplot2grid(tamanho_da_grade, posição_inicial, colspan=..., rowspan=...)
ax1 = plt.subplot2grid((3, 3), (0, 0), colspan=3)
ax2 = plt.subplot2grid((3, 3), (1, 0), colspan=2)
ax3 = plt.subplot2grid((3,3), (1, 2), rowspan=2)
ax4 = plt.subplot2grid((3,3), (2, 0))
ax5 = plt.subplot2grid((3,3), (2, 1))
```

##**Scatter plot

```python
N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
# area dos circulos usada no grafico
area = np.pi*(20 * np.random.rand(N))**2 # 0 to 15 point radii
```

```python
plt.figure(figsize=(14, 6))

plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap='Spectral')
plt.colorbar()

plt.show()
```

```python
fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(1, 2, 1)
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap='Pastel1')
plt.colorbar()

ax2 = fig.add_subplot(1, 2, 2)
plt.scatter(x, y, s=area, c=colors, alpha=0.5, cmap='Pastel2')
plt.colorbar()

plt.show()
```

##**Histograms**

```python
values = np.random.randn(100)
```

```python
plt.subplots(figsize=(12, 6))

plt.hist(values, bins=100, alpha=0.8,
         histtype="bar", color='steelblue',
         edgecolor='green')
plt.xlim(xmin=-5, xmax=5)

plt.show()
```

```python
fig.savefig('hist.png')
```

```python
##**KDE (kernel density estimation)
```

```python
from scipy import stats

density = stats.kde.gaussian_kde(values)
```

```python
plt.subplots(figsize=(12, 6))

values2 = np.linspace(min(values)-10, max(values)+10, 100)

plt.plot(values2, density(values2), color="#FF7F00")
plt.fill_between(values2, density(values2), alpha=0.5, color="#FF7F00")
plt.xlim(xmin=-5, xmax=5)

plt.show()
```

##**Combine plots**

```python
plt.subplots(figsize=(12, 6))

plt.hist(values, bins=100, alpha=0.8, density=1,
         histtype='bar', color='steelblue',
         edgecolor='green')
plt.plot(values2, density(values2), color="#ff7f00", linewidth=3.0)
plt.xlim(xmin=-5, xmax=5)  #limite do eixo x

plt.show()
```

##**Bar plots**

```python
# Gera o array 1D (5 elementos)
Y = np.random.rand(1, 5)[0]
Y2 = np.random.rand(1, 5)[0]
```

```python
plt.figure(figsize=(12, 4))

# Largura das barras
barWidth = 0.5

# Posição X: np.arange(len(Y)) gera array([0, 1, 2, 3, 4])
# Posição Y: passa o próprio array Y
plt.bar(np.arange(len(Y)), Y, width=barWidth, color="#00b894")

plt.show()
```

Also can be stacked bars, and add a legend to the plot:

```python
plt.figure(figsize=(12, 4))

barWidth = 0.5
plt.bar(np.arange(len(Y)), Y, width=barWidth, color="#00b894", label='Label Y')
plt.bar(np.arange(len(Y2)), Y2, width=barWidth, color="#e17055", bottom=Y, label='Label Y2')

plt.legend()

plt.show()
```
